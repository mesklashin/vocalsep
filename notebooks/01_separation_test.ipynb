# 🎵 Phase 2 — Separation Test & Verification

This notebook verifies the Demucs integration and allows you to visual/audio check the results of the batch separation.

```python
import os
import sys
from pathlib import Path
import librosa
import librosa.display
import matplotlib.pyplot as plt
import IPython.display as ipd

# Set project root
parent_dir = str(Path(os.getcwd()).parent)
if parent_dir not in sys.path:
    sys.path.append(parent_dir)

import config
print(f"Project Root: {config.BASE_DIR}")
```

## 1. Check Available Separated Songs

```python
separated_dir = Path(config.SEPARATED_DATA_DIR)
models = [d.name for d in separated_dir.iterdir() if d.is_dir()]
print(f"Models found: {models}")

for model in models:
    songs = [d.name for d in (separated_dir / model).iterdir() if d.is_dir()]
    print(f"  {model}: {songs}")
```

## 2. Audio Preview (Pick a song and model)

```python
# --- CONFIGURATION ---
SELECTED_MODEL = "htdemucs" 
SELECTED_SONG = "test_song" # Change this to an actual song folder name
# ---------------------

song_path = Path(config.SEPARATED_DATA_DIR) / SELECTED_MODEL / SELECTED_SONG

if song_path.exists():
    print(f"Vocals:")
    ipd.display(ipd.Audio(song_path / "vocals.wav"))
    
    print(f"Other (Instrumental):")
    ipd.display(ipd.Audio(song_path / "other.wav"))
else:
    print(f"Path not found: {song_path}. Have you run batch_separate.py?")
```

## 3. Waveform Comparison

```python
def plot_stems(song_dir):
    stems = ["vocals", "drums", "bass", "other"]
    plt.figure(figsize=(15, 10))
    
    for i, stem in enumerate(stems):
        path = Path(song_dir) / f"{stem}.wav"
        if path.exists():
            y, sr = librosa.load(path, duration=30) # Load first 30s
            plt.subplot(4, 1, i+1)
            librosa.display.waveshow(y, sr=sr)
            plt.title(stem.capitalize())
    
    plt.tight_layout()
    plt.show()

if song_path.exists():
    plot_stems(song_path)
```
